```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ── H0 TENSION DATA ──────────────────────────────────────────────────
# Internal rulers (distance ladder) — subject to η_opt trace defect
h0_internal = [
    ("SH0ES (Riess 2022)",        73.04, 1.04, "ladder"),
    ("HST TRGB (Freedman 2021)",  69.80, 1.70, "ladder"),
    ("Megamaser Cosmology",       73.90, 3.00, "ladder"),
    ("Surface Brightness Fluct.", 73.70, 2.40, "ladder"),
]

# External rulers (BAO/CMB) — geometric baseline
h0_external = [
    ("Planck 2018 (CMB)",         67.36, 0.54, "geometric"),
    ("DESI DR1 (BAO)",            68.52, 0.62, "geometric"),
    ("DES Y3 + BAO",              68.20, 1.10, "geometric"),
    ("SPT-3G + BAO",              68.80, 1.50, "geometric"),
]

# Oros Fundus H0 prediction
h0_OF = 68.6  # km/s/Mpc — derived, not fitted

# ── S8 TENSION DATA ──────────────────────────────────────────────────
# High S8 — CMB-inferred (internal to ΛCDM)
s8_high = [
    ("Planck 2018 (CMB)",         0.834, 0.016, "cmb"),
    ("ACT DR4",                   0.840, 0.030, "cmb"),
]

# Low S8 — weak lensing surveys (direct geometric measurement)
s8_low = [
    ("KiDS-1000",                 0.766, 0.020, "lensing"),
    ("DES Y3",                    0.776, 0.017, "lensing"),
    ("HSC Y3",                    0.775, 0.025, "lensing"),
    ("KiDS + BOSS + 2dFLenS",     0.742, 0.035, "lensing"),
]

# Oros Fundus S8 prediction 
# Planck value suppressed by modular correction (~2.5% suppression)
s8_OF = 0.834 * (1 - 0.025) 

print("Data loaded successfully.")
print(f"Oros Fundus H₀: {h0_OF} km/s/Mpc")
print(f"Oros Fundus S₈:  {s8_OF:.3f}")
```

```python
fig_dashboard = make_subplots(
    rows=2, cols=1,
    subplot_titles=(
        'H₀ Tension — Two Rulers, Two Answers',
        'S₈ Tension — Structure Growth Suppression'
    ),
    vertical_spacing=0.12,
    row_heights=[0.55, 0.45]
)

# ── H0 PANEL (row 1) ──────────────────────────────────────────────────────────
for i, (name, h0, err, kind) in enumerate(h0_external):
    fig_dashboard.add_trace(go.Scatter(
        x=[h0], y=[len(h0_internal) + 1 + i],
        mode='markers', name=name,
        marker=dict(color='#4FC3F7', size=10, symbol='square'),
        error_x=dict(type='data', array=[err], color='#4FC3F7',
                     thickness=2, width=5),
        legendgroup='h0', showlegend=True
    ), row=1, col=1)

for i, (name, h0, err, kind) in enumerate(h0_internal):
    fig_dashboard.add_trace(go.Scatter(
        x=[h0], y=[i],
        mode='markers', name=name,
        marker=dict(color='#EF5350', size=10, symbol='circle'),
        error_x=dict(type='data', array=[err], color='#EF5350',
                     thickness=2, width=5),
        legendgroup='h0', showlegend=True
    ), row=1, col=1)

# ── S8 PANEL (row 2) ──────────────────────────────────────────────────────────
for i, (name, s8, err, kind) in enumerate(s8_high):
    fig_dashboard.add_trace(go.Scatter(
        x=[s8], y=[len(s8_low) + 1 + i],
        mode='markers', name=name,
        marker=dict(color='#FF7043', size=10, symbol='diamond'),
        error_x=dict(type='data', array=[err], color='#FF7043',
                     thickness=2, width=5),
        legendgroup='s8', showlegend=True
    ), row=2, col=1)

for i, (name, s8, err, kind) in enumerate(s8_low):
    fig_dashboard.add_trace(go.Scatter(
        x=[s8], y=[i],
        mode='markers', name=name,
        marker=dict(color='#CE93D8', size=10, symbol='circle'),
        error_x=dict(type='data', array=[err], color='#CE93D8',
                     thickness=2, width=5),
        legendgroup='s8', showlegend=True
    ), row=2, col=1)

# ── OF LINES ──────────────────────────────────────────────────────────────────
fig_dashboard.add_vline(x=h0_OF, line_dash='dash',
                        line_color='#69F0AE', line_width=2, row=1, col=1)
fig_dashboard.add_vline(x=s8_OF, line_dash='dash',
                        line_color='#69F0AE', line_width=2, row=2, col=1)

# ── SHADED REGIONS ────────────────────────────────────────────────────────────
fig_dashboard.add_vrect(x0=71.5, x1=75.5, fillcolor='rgba(239,83,80,0.08)', line_width=0, row=1, col=1)
fig_dashboard.add_vrect(x0=66.5, x1=70.0, fillcolor='rgba(79,195,247,0.08)', line_width=0, row=1, col=1)
fig_dashboard.add_vrect(x0=0.820, x1=0.875, fillcolor='rgba(255,112,67,0.08)', line_width=0, row=2, col=1)
fig_dashboard.add_vrect(x0=0.730, x1=0.800, fillcolor='rgba(206,147,216,0.08)', line_width=0, row=2, col=1)

# ── Y AXIS LABELS ─────────────────────────────────────────────────────────────
h0_labels = [m[0] for m in h0_internal] + [''] + [m[0] for m in h0_external]
s8_labels = [m[0] for m in s8_low] + [''] + [m[0] for m in s8_high]

fig_dashboard.update_yaxes(
    tickmode='array',
    tickvals=list(range(len(h0_internal))) + [len(h0_internal)] + [len(h0_internal) + 1 + i for i in range(len(h0_external))],
    ticktext=h0_labels, showgrid=False, row=1, col=1
)
fig_dashboard.update_yaxes(
    tickmode='array',
    tickvals=list(range(len(s8_low))) + [len(s8_low)] + [len(s8_low) + 1 + i for i in range(len(s8_high))],
    ticktext=s8_labels, showgrid=False, row=2, col=1
)

fig_dashboard.update_xaxes(title='H₀ (km/s/Mpc)', range=[64, 78], showgrid=True, gridcolor='rgba(255,255,255,0.08)', row=1, col=1)
fig_dashboard.update_xaxes(title='S₈ = σ₈ √(Ωₘ/0.3)', range=[0.70, 0.88], showgrid=True, gridcolor='rgba(255,255,255,0.08)', row=2, col=1)

# ── OF ANNOTATIONS ────────────────────────────────────────────────────────────
fig_dashboard.add_annotation(
    x=h0_OF, y=len(h0_internal) + len(h0_external) + 1,
    text=f'OF: {h0_OF} km/s/Mpc', showarrow=True, arrowhead=2,
    arrowcolor='#69F0AE', font=dict(color='#69F0AE', size=10),
    bgcolor='rgba(0,0,0,0.6)', ax=60, ay=0, row=1, col=1
)
fig_dashboard.add_annotation(
    x=s8_OF, y=len(s8_low) + len(s8_high) + 1,
    text=f'OF: {s8_OF:.3f}', showarrow=True, arrowhead=2,
    arrowcolor='#69F0AE', font=dict(color='#69F0AE', size=10),
    bgcolor='rgba(0,0,0,0.6)', ax=60, ay=0, row=2, col=1
)

# ── LAYOUT ────────────────────────────────────────────────────────────────────
fig_dashboard.update_layout(
    title=dict(
        text='Cosmological Tension Resolution — Oros Fundus Framework<br>'
             '<sup>η_opt trace defect systematically offsets internal rulers from geometric baseline</sup>',
        font=dict(size=15), x=0.5
    ),
    template='plotly_dark', height=820,
    legend=dict(x=1.01, y=1.0, bgcolor='rgba(0,0,0,0.5)', bordercolor='rgba(255,255,255,0.15)', borderwidth=1, font=dict(size=10)),
    margin=dict(l=200, r=220, t=100, b=60)
)

fig_dashboard.show()
```
